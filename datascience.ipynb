```python
import pandas as pd
import numpy as np
import glob
import os

# Set display options for easier viewing
pd.set_option('display.max_columns', 50)
pd.set_option('display.width', 150)
```

week 1

```python
# Update this path to where you saved the downloaded CRMLSSold files
data_folder = "/Users/rucha/Documents/IDX/CRMLSSold_data"
file_paths = glob.glob(os.path.join(data_folder, "CRMLSSold*.csv"))

print(f"Found {len(file_paths)} files")
for f in file_paths:
    print(os.path.basename(f))

# Load one file to check structure and columns before combining
sample_df = pd.read_csv(file_paths[0])
print("\nShape:", sample_df.shape)
print("\nColumns:", sample_df.columns.tolist())
sample_df.head()
```

week 2

```python
import matplotlib.pyplot as plt

# Combine all monthly files into one dataframe
df_list = [pd.read_csv(f) for f in file_paths]
df_raw = pd.concat(df_list, ignore_index=True)
print("Combined shape:", df_raw.shape)

# Filter to the scope required by the task
df = df_raw[
    (df_raw['PropertyType'] == 'Residential') &
    (df_raw['PropertySubType'] == 'SingleFamilyResidence')
].copy()
print(f"Rows before filter: {len(df_raw)}")
print(f"Rows after filter: {len(df)}")

# Missing values check
print("\nMissing values per column:")
print(df.isnull().sum().sort_values(ascending=False).head(20))

# ClosePrice distribution
print("\nClosePrice summary:")
print(df['ClosePrice'].describe())

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
df['ClosePrice'].hist(bins=50, ax=axes[0,0])
axes[0,0].set_title('Close Price')

df['LivingArea'].hist(bins=50, ax=axes[0,1])
axes[0,1].set_title('Living Area')

df['BedroomsTotal'].value_counts().sort_index().plot(kind='bar', ax=axes[1,0])
axes[1,0].set_title('Bedrooms')

df['LotSizeSquareFeet'].hist(bins=50, ax=axes[1,1])
axes[1,1].set_title('Lot Size')

plt.tight_layout()
plt.show()

# Quick outlier check
print("\nRows with ClosePrice <= 0:", (df['ClosePrice'] <= 0).sum())
print("Rows with LivingArea <= 0:", (df['LivingArea'] <= 0).sum())
```
