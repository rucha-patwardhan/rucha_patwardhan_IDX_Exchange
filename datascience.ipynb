```python
# Filter to the required scope
df = df_final[df_final['PropertySubType'] == 'SingleFamilyResidence'].copy()
print(f"Rows before filter: {len(df_final)}")
print(f"Rows after filter: {len(df)}")

# Convert CloseDate to datetime for time based splitting
df['CloseDate'] = pd.to_datetime(df['CloseDate'])

# Use the most recent month as test, prior months as train
df['close_year_month'] = df['CloseDate'].dt.to_period('M')
most_recent_month = df['close_year_month'].max()
print("\nMost recent month (test set):", most_recent_month)

test_df = df[df['close_year_month'] == most_recent_month].copy()

# Try a training window of X months before the test month, start with X = 6
X = 6
train_months = pd.period_range(end=most_recent_month - 1, periods=X, freq='M')
train_df = df[df['close_year_month'].isin(train_months)].copy()

print(f"\nTrain set (X={X} months): {len(train_df)} rows, {train_df['close_year_month'].min()} to {train_df['close_year_month'].max()}")
print(f"Test set: {len(test_df)} rows")
```

Week 4

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Pick a simple set of numeric features for the baseline
features = ['LivingArea', 'BedroomsTotal', 'BathroomsTotalInteger', 'LotSizeSquareFeet', 'YearBuilt']
target = 'ClosePrice'

# Drop rows with missing values in these columns for the baseline
train_model_df = train_df[features + [target]].dropna()
test_model_df = test_df[features + [target]].dropna()

print(f"Train rows after dropping missing: {len(train_model_df)}")
print(f"Test rows after dropping missing: {len(test_model_df)}")

X_train = train_model_df[features]
y_train = train_model_df[target]
X_test = test_model_df[features]
y_test = test_model_df[target]

# Train baseline linear regression
model = LinearRegression()
model.fit(X_train, y_train)

# Evaluate
train_preds = model.predict(X_train)
test_preds = model.predict(X_test)

train_r2 = r2_score(y_train, train_preds)
test_r2 = r2_score(y_test, test_preds)

print(f"\nTrain R²: {train_r2:.4f}")
print(f"Test R²: {test_r2:.4f}")

# Look at coefficients to understand what's driving predictions
coef_df = pd.DataFrame({'Feature': features, 'Coefficient': model.coef_})
print("\nCoefficients:")
print(coef_df)
```

Baseline Linear Regression, Test R² = 0.3076
Features used: LivingArea, BedroomsTotal, BathroomsTotalInteger, LotSizeSquareFeet, YearBuilt
~1,000 train rows and ~130 test rows dropped due to missing values in these features
Negative coefficients on Bedrooms and YearBuilt likely reflect multicollinearity, not causal relationships

Week 5

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

# Decision Tree
tree_model = DecisionTreeRegressor(max_depth=10, random_state=42)
tree_model.fit(X_train, y_train)

tree_train_preds = tree_model.predict(X_train)
tree_test_preds = tree_model.predict(X_test)

tree_train_r2 = r2_score(y_train, tree_train_preds)
tree_test_r2 = r2_score(y_test, tree_test_preds)

print("Decision Tree:")
print(f"Train R²: {tree_train_r2:.4f}")
print(f"Test R²: {tree_test_r2:.4f}")

# Random Forest
rf_model = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

rf_train_preds = rf_model.predict(X_train)
rf_test_preds = rf_model.predict(X_test)

rf_train_r2 = r2_score(y_train, rf_train_preds)
rf_test_r2 = r2_score(y_test, rf_test_preds)

print("\nRandom Forest:")
print(f"Train R²: {rf_train_r2:.4f}")
print(f"Test R²: {rf_test_r2:.4f}")

# Compare all three models side by side
comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Decision Tree', 'Random Forest'],
    'Train R²': [train_r2, tree_train_r2, rf_train_r2],
    'Test R²': [test_r2, tree_test_r2, rf_test_r2]
})
print("\nModel Comparison:")
print(comparison)

# Feature importance from Random Forest
importance_df = pd.DataFrame({
    'Feature': features,
    'Importance': rf_model.feature_importances_
}).sort_values('Importance', ascending=False)
print("\nRandom Forest Feature Importance:")
print(importance_df)
```

```python
import geopandas as gpd
from shapely.geometry import Point

school_district_url = "https://data.ca.gov/dataset/california-school-district-areas-2024-25/resource/7dfaf005-58eb-45db-93b1-7aff091b2172/download"
districts = gpd.read_file(school_district_url)
print("District boundaries loaded:", districts.shape)
print(districts.columns.tolist())
districts.head()
```

```python
print("District CRS:", districts.crs)
```

```python
# Only properties with valid coordinates can be spatially joined
df_geo = df[df['Latitude'].notna() & df['Longitude'].notna()].copy()
print(f"Rows with valid coordinates: {len(df_geo)} out of {len(df)}")

# Convert to a GeoDataFrame using standard lat/long (EPSG:4326)
df_geo['geometry'] = df_geo.apply(lambda row: Point(row['Longitude'], row['Latitude']), axis=1)
gdf_properties = gpd.GeoDataFrame(df_geo, geometry='geometry', crs='EPSG:4326')

# Reproject districts to match, using whatever their actual CRS is
districts_matched = districts.to_crs('EPSG:4326')

# Spatial join: match each property point to the district polygon it falls inside
joined = gpd.sjoin(gdf_properties, districts_matched, how='left', predicate='within')

print("\nJoined shape:", joined.shape)
print("Rows matched to a district:", joined['DistrictNa'].notna().sum())
print("Rows unmatched:", joined['DistrictNa'].isna().sum())
```

check

```python
# Check for duplicate properties in the join (same ListingKey matched to multiple districts)
dup_check = joined['ListingKey'].value_counts()
print("Properties matched to more than one district:", (dup_check > 1).sum())
print("\nExample of a duplicated property:")
example_key = dup_check[dup_check > 1].index[0]
print(joined[joined['ListingKey'] == example_key][['ListingKey', 'DistrictNa', 'DistrictTy']])
```

```python
# Deduplicate: prefer Unified district type, otherwise take the first match
joined_sorted = joined.sort_values(by='DistrictTy', key=lambda x: x != 'Unified')
joined_dedup = joined_sorted.drop_duplicates(subset='ListingKey', keep='first')

print("Deduplicated shape:", joined_dedup.shape)
print("Original valid coordinate rows:", len(df_geo))
```

Merging

```python
# Merge the district name back into the main dataframe
district_map = joined_dedup.set_index('ListingKey')['DistrictNa']
df['school_district'] = df['ListingKey'].map(district_map)

print("Rows with school_district assigned:", df['school_district'].notna().sum())
print("Rows missing (no coordinates or no match):", df['school_district'].isna().sum())
```

Rebuilding

```python
# Rebuild train/test with all Week 6 features included
test_df = df[df['close_year_month'] == most_recent_month].copy()
train_df = df[df['close_year_month'].isin(train_months)].copy()

print(f"Train: {len(train_df)} rows")
print(f"Test: {len(test_df)} rows")

# Updated feature list including Week 6 engineered features
# school_district is categorical, so encode it for now with a simple approach
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
# Fit on train, handle unseen categories in test by mapping to a placeholder
train_df['school_district_encoded'] = le.fit_transform(train_df['school_district'].astype(str))

# Map test set using the same encoder, unseen districts get -1
district_map_encoded = dict(zip(le.classes_, le.transform(le.classes_)))
test_df['school_district_encoded'] = test_df['school_district'].astype(str).map(district_map_encoded).fillna(-1)

features_v2 = ['LivingArea', 'BedroomsTotal', 'BathroomsTotalInteger', 'LotSizeSquareFeet',
               'YearBuilt', 'bed_bath_ratio', 'property_age', 'school_district_encoded']
target = 'ClosePrice'

train_model_df = train_df[features_v2 + [target]].dropna()
test_model_df = test_df[features_v2 + [target]].dropna()

print(f"\nTrain rows after dropping missing: {len(train_model_df)}")
print(f"Test rows after dropping missing: {len(test_model_df)}")

X_train_v2 = train_model_df[features_v2]
y_train_v2 = train_model_df[target]
X_test_v2 = test_model_df[features_v2]
y_test_v2 = test_model_df[target]

# Retrain Random Forest with the expanded feature set
rf_model_v2 = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42, n_jobs=-1)
rf_model_v2.fit(X_train_v2, y_train_v2)

rf_v2_train_preds = rf_model_v2.predict(X_train_v2)
rf_v2_test_preds = rf_model_v2.predict(X_test_v2)

rf_v2_train_r2 = r2_score(y_train_v2, rf_v2_train_preds)
rf_v2_test_r2 = r2_score(y_test_v2, rf_v2_test_preds)

print(f"\nRandom Forest v2 (with Week 6 features):")
print(f"Train R²: {rf_v2_train_r2:.4f}")
print(f"Test R²: {rf_v2_test_r2:.4f}")

# Compare old vs new feature set
comparison_v2 = pd.DataFrame({
    'Model': ['Random Forest (Week 5)', 'Random Forest (Week 6, with new features)'],
    'Train R²': [rf_train_r2, rf_v2_train_r2],
    'Test R²': [rf_test_r2, rf_v2_test_r2]
})
print("\nOld vs New Feature Set Comparison:")
print(comparison_v2)

# Updated feature importance
importance_df_v2 = pd.DataFrame({
    'Feature': features_v2,
    'Importance': rf_model_v2.feature_importances_
}).sort_values('Importance', ascending=False)
print("\nUpdated Feature Importance:")
print(importance_df_v2)
```

Week 6 feature engineering (bed/bath ratio, property age, school district spatial join) improved Random Forest Test R² from 0.4138 to 0.4826. School district and property age emerged as strong predictors, validating the geographic feature engineering effort.

Week 7- xgboost installed

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train_v2, y_train_v2)

xgb_train_preds = xgb_model.predict(X_train_v2)
xgb_test_preds = xgb_model.predict(X_test_v2)

xgb_train_r2 = r2_score(y_train_v2, xgb_train_preds)
xgb_test_r2 = r2_score(y_test_v2, xgb_test_preds)

print("XGBoost (initial):")
print(f"Train R²: {xgb_train_r2:.4f}")
print(f"Test R²: {xgb_test_r2:.4f}")
```

```python
from itertools import product

param_grid = {
    'max_depth': [4, 6, 8],
    'learning_rate': [0.05, 0.1],
    'n_estimators': [200, 400]
}

results = []

for depth, lr, n_est in product(param_grid['max_depth'], param_grid['learning_rate'], param_grid['n_estimators']):
    model = XGBRegressor(
        n_estimators=n_est,
        max_depth=depth,
        learning_rate=lr,
        random_state=42,
        n_jobs=-1
    )
    model.fit(X_train_v2, y_train_v2)
    train_preds = model.predict(X_train_v2)
    test_preds = model.predict(X_test_v2)
    train_r2 = r2_score(y_train_v2, train_preds)
    test_r2 = r2_score(y_test_v2, test_preds)
    results.append({
        'max_depth': depth,
        'learning_rate': lr,
        'n_estimators': n_est,
        'train_r2': train_r2,
        'test_r2': test_r2,
        'gap': train_r2 - test_r2
    })

results_df = pd.DataFrame(results).sort_values('test_r2', ascending=False)
print(results_df)
```

```python
# Rerun all models fresh to avoid variable name collisions

# Linear Regression
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
lr_train_r2 = r2_score(y_train, lr_model.predict(X_train))
lr_test_r2 = r2_score(y_test, lr_model.predict(X_test))

# Decision Tree
dt_model = DecisionTreeRegressor(max_depth=10, random_state=42)
dt_model.fit(X_train, y_train)
dt_train_r2 = r2_score(y_train, dt_model.predict(X_train))
dt_test_r2 = r2_score(y_test, dt_model.predict(X_test))

# Random Forest (Week 5 features)
rf5_model = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42, n_jobs=-1)
rf5_model.fit(X_train, y_train)
rf5_train_r2 = r2_score(y_train, rf5_model.predict(X_train))
rf5_test_r2 = r2_score(y_test, rf5_model.predict(X_test))

# Random Forest (Week 6 features)
rf6_model = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42, n_jobs=-1)
rf6_model.fit(X_train_v2, y_train_v2)
rf6_train_r2 = r2_score(y_train_v2, rf6_model.predict(X_train_v2))
rf6_test_r2 = r2_score(y_test_v2, rf6_model.predict(X_test_v2))

# XGBoost tuned (Week 7)
xgb_final = XGBRegressor(max_depth=6, learning_rate=0.05, n_estimators=400, random_state=42, n_jobs=-1)
xgb_final.fit(X_train_v2, y_train_v2)
xgb_train_r2 = r2_score(y_train_v2, xgb_final.predict(X_train_v2))
xgb_test_r2 = r2_score(y_test_v2, xgb_final.predict(X_test_v2))

full_comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Decision Tree', 'Random Forest (Week 5)',
              'Random Forest (Week 6, new features)', 'XGBoost (tuned, Week 7)'],
    'Train R²': [lr_train_r2, dt_train_r2, rf5_train_r2, rf6_train_r2, xgb_train_r2],
    'Test R²': [lr_test_r2, dt_test_r2, rf5_test_r2, rf6_test_r2, xgb_test_r2]
})
print(full_comparison)
```

Week 8

```python
# Predictions from the tuned XGBoost model (Week 7)
y_pred = xgb_final.predict(X_test_v2)

# MAPE: Mean Absolute Percentage Error
def mape(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

# MdAPE: Median Absolute Percentage Error (more robust to outliers than MAPE)
def mdape(y_true, y_pred):
    return np.median(np.abs((y_true - y_pred) / y_true)) * 100

test_mape = mape(y_test_v2, y_pred)
test_mdape = mdape(y_test_v2, y_pred)

print(f"Test MAPE: {test_mape:.2f}%")
print(f"Test MdAPE: {test_mdape:.2f}%")
print(f"Test R²: {r2_score(y_test_v2, y_pred):.4f}")
```

```python
# Build a results dataframe with actual, predicted, and error
results_df = pd.DataFrame({
    'actual': y_test_v2.values,
    'predicted': y_pred
})
results_df['abs_pct_error'] = np.abs((results_df['actual'] - results_df['predicted']) / results_df['actual']) * 100

# Create price bands
bins = [0, 300000, 500000, 750000, 1000000, 1500000, 2000000, np.inf]
labels = ['<300k', '300k-500k', '500k-750k', '750k-1M', '1M-1.5M', '1.5M-2M', '2M+']
results_df['price_band'] = pd.cut(results_df['actual'], bins=bins, labels=labels)

# Summarize by price band
band_summary = results_df.groupby('price_band', observed=True).agg(
    count=('actual', 'size'),
    mean_mape=('abs_pct_error', 'mean'),
    median_mape=('abs_pct_error', 'median')
).round(2)
print(band_summary)
```

```python
# Investigate the worst offenders in the <300k band
low_band = results_df[results_df['price_band'] == '<300k'].sort_values('abs_pct_error', ascending=False)
print(low_band.head(10))

# Check for suspiciously low actual prices
print("\nRows with actual < $10,000:")
print(results_df[results_df['actual'] < 10000])
```

fixing

```python
# Recalculate MAPE excluding clear data errors (e.g. price below a reasonable floor)
clean_results = results_df[results_df['actual'] >= 50000].copy()

clean_mape = clean_results['abs_pct_error'].mean()
clean_mdape = clean_results['abs_pct_error'].median()

print(f"Rows excluded as likely data errors: {len(results_df) - len(clean_results)}")
print(f"Cleaned MAPE (actual >= $50,000): {clean_mape:.2f}%")
print(f"Cleaned MdAPE (actual >= $50,000): {clean_mdape:.2f}%")
```

Summary

```python
metrics_summary = pd.DataFrame({
    'Metric': ['R²', 'MAPE (raw)', 'MAPE (cleaned, excl. data errors)', 'MdAPE'],
    'Value': [f"{r2_score(y_test_v2, y_pred):.4f}", f"{test_mape:.2f}%", f"{clean_mape:.2f}%", f"{test_mdape:.2f}%"]
})
print(metrics_summary)

# Save the price band breakdown and metrics summary for your documentation
band_summary.to_csv('price_band_evaluation.csv')
metrics_summary.to_csv('metrics_summary.csv', index=False)
print("\nSaved metrics_summary.csv and price_band_evaluation.csv")
```

- Test R² = 0.6595, the tuned XGBoost model explains about 66% of the variance in close price
- MdAPE = 15.61%, the most reliable headline metric, predictions are typically within about 16% of actual price
- Raw MAPE was severely distorted by 2 data entry errors (one row had an actual close price of $1.75); after excluding those, cleaned MAPE = 25.39%
- The gap between MAPE (25.39%) and MdAPE (15.61%) is itself worth noting, it shows a subset of predictions have large errors that pull the average up, worth investigating further, likely concentrated in the very low and very high price bands as the earlier breakdown showed
- Performance by price band: strongest in the 750k-1.5M range (MdAPE ~12%), weakest at the extremes, under 300k (after removing bad data) and 2M+ (MdAPE ~30%)
