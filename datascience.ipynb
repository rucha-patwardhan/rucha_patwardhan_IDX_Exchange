```python
df_final = pd.read_csv("sold_final.csv")
print("Shape:", df_final.shape)
print("\nColumns:", df_final.columns.tolist())
df_final.head()
```

```python
print(df_final['PropertySubType'].value_counts())
print("\nDate range:", df_final['CloseDate'].min(), "to", df_final['CloseDate'].max())
```

```python
# Filter to the required scope
df = df_final[df_final['PropertySubType'] == 'SingleFamilyResidence'].copy()
print(f"Rows before filter: {len(df_final)}")
print(f"Rows after filter: {len(df)}")

# Convert CloseDate to datetime for time based splitting
df['CloseDate'] = pd.to_datetime(df['CloseDate'])

# Use the most recent month as test, prior months as train
df['close_year_month'] = df['CloseDate'].dt.to_period('M')
most_recent_month = df['close_year_month'].max()
print("\nMost recent month (test set):", most_recent_month)

test_df = df[df['close_year_month'] == most_recent_month].copy()

# Try a training window of X months before the test month, start with X = 6
X = 6
train_months = pd.period_range(end=most_recent_month - 1, periods=X, freq='M')
train_df = df[df['close_year_month'].isin(train_months)].copy()

print(f"\nTrain set (X={X} months): {len(train_df)} rows, {train_df['close_year_month'].min()} to {train_df['close_year_month'].max()}")
print(f"Test set: {len(test_df)} rows")
```

Week 4

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Pick a simple set of numeric features for the baseline
features = ['LivingArea', 'BedroomsTotal', 'BathroomsTotalInteger', 'LotSizeSquareFeet', 'YearBuilt']
target = 'ClosePrice'

# Drop rows with missing values in these columns for the baseline
train_model_df = train_df[features + [target]].dropna()
test_model_df = test_df[features + [target]].dropna()

print(f"Train rows after dropping missing: {len(train_model_df)}")
print(f"Test rows after dropping missing: {len(test_model_df)}")

X_train = train_model_df[features]
y_train = train_model_df[target]
X_test = test_model_df[features]
y_test = test_model_df[target]

# Train baseline linear regression
model = LinearRegression()
model.fit(X_train, y_train)

# Evaluate
train_preds = model.predict(X_train)
test_preds = model.predict(X_test)

train_r2 = r2_score(y_train, train_preds)
test_r2 = r2_score(y_test, test_preds)

print(f"\nTrain R²: {train_r2:.4f}")
print(f"Test R²: {test_r2:.4f}")

# Look at coefficients to understand what's driving predictions
coef_df = pd.DataFrame({'Feature': features, 'Coefficient': model.coef_})
print("\nCoefficients:")
print(coef_df)
```

Baseline Linear Regression, Test R² = 0.3076
Features used: LivingArea, BedroomsTotal, BathroomsTotalInteger, LotSizeSquareFeet, YearBuilt
~1,000 train rows and ~130 test rows dropped due to missing values in these features
Negative coefficients on Bedrooms and YearBuilt likely reflect multicollinearity, not causal relationships

week 5

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

# Decision Tree
tree_model = DecisionTreeRegressor(max_depth=10, random_state=42)
tree_model.fit(X_train, y_train)

tree_train_preds = tree_model.predict(X_train)
tree_test_preds = tree_model.predict(X_test)

tree_train_r2 = r2_score(y_train, tree_train_preds)
tree_test_r2 = r2_score(y_test, tree_test_preds)

print("Decision Tree:")
print(f"Train R²: {tree_train_r2:.4f}")
print(f"Test R²: {tree_test_r2:.4f}")

# Random Forest
rf_model = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

rf_train_preds = rf_model.predict(X_train)
rf_test_preds = rf_model.predict(X_test)

rf_train_r2 = r2_score(y_train, rf_train_preds)
rf_test_r2 = r2_score(y_test, rf_test_preds)

print("\nRandom Forest:")
print(f"Train R²: {rf_train_r2:.4f}")
print(f"Test R²: {rf_test_r2:.4f}")

# Compare all three models side by side
comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Decision Tree', 'Random Forest'],
    'Train R²': [train_r2, tree_train_r2, rf_train_r2],
    'Test R²': [test_r2, tree_test_r2, rf_test_r2]
})
print("\nModel Comparison:")
print(comparison)

# Feature importance from Random Forest
importance_df = pd.DataFrame({
    'Feature': features,
    'Importance': rf_model.feature_importances_
}).sort_values('Importance', ascending=False)
print("\nRandom Forest Feature Importance:")
print(importance_df)
```
