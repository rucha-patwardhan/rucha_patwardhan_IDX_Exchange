```python
files = sorted([f for f in os.listdir('raw') if f.startswith('CRMLSSold') and f.endswith('.csv')])
sold = pd.concat([pd.read_csv(f'raw/{f}') for f in files], ignore_index=True)
sold = sold[sold['PropertyType'] == 'Residential']
print(sold.shape)
```

```python
print(sold.shape)
print(sold.dtypes)
```

```python
print(sold['PropertyType'].unique())
```

```python
missing = sold.isnull().sum()
missing_pct = (missing / len(sold) * 100).round(2)
missing_report = pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
missing_report = missing_report[missing_report['missing_count'] > 0].sort_values('missing_pct', ascending=False)
print(missing_report)
```

```python
cols = ['ClosePrice', 'ListPrice', 'OriginalListPrice', 'LivingArea', 'DaysOnMarket', 'BedroomsTotal', 'BathroomsTotalInteger']
print(sold[cols].describe())
```

```python
url = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=MORTGAGE30US"
mortgage = pd.read_csv(url)
print(mortgage.columns.tolist())  # see exact column names
print(mortgage.head())
```

```python
mortgage.columns = ['date', 'rate_30yr_fixed']
mortgage['date'] = pd.to_datetime(mortgage['date'])
mortgage['year_month'] = mortgage['date'].dt.to_period('M')
mortgage_monthly = mortgage.groupby('year_month')['rate_30yr_fixed'].mean().reset_index()
print(mortgage_monthly.tail())
```

```python
sold['year_month'] = pd.to_datetime(sold['CloseDate']).dt.to_period('M')
sold = sold.merge(mortgage_monthly, on='year_month', how='left')
print(sold['rate_30yr_fixed'].isnull().sum())
print(sold[['CloseDate', 'year_month', 'ClosePrice', 'rate_30yr_fixed']].head())
```

```python
sold.to_csv('sold_combined.csv', index=False)
print("Saved sold_combined.csv with mortgage rates!")
```

```python
os.chdir('/Users/ruchapatwardhan/Desktop/IDX Internship')
sold = pd.read_csv('sold_combined.csv', low_memory=False)
print(sold.shape)
```

```python
# Drop columns that are completely empty
cols_before = sold.shape[1]
sold = sold.dropna(axis=1, how='all')
cols_after = sold.shape[1]
print(f"Columns before: {cols_before}")
print(f"Columns after: {cols_after}")
print(f"Columns dropped: {cols_before - cols_after}")
```

```python
date_cols = ['CloseDate', 'ListingContractDate', 'PurchaseContractDate', 'ContractStatusChangeDate']
for col in date_cols:
    if col in sold.columns:
        sold[col] = pd.to_datetime(sold[col], errors='coerce')
        print(f"{col} converted successfully")
```

```python
sold['invalid_close_price'] = sold['ClosePrice'] <= 0
sold['invalid_living_area'] = sold['LivingArea'] <= 0
sold['invalid_days_on_market'] = sold['DaysOnMarket'] < 0

print(f"Invalid ClosePrice: {sold['invalid_close_price'].sum()}")
print(f"Invalid LivingArea: {sold['invalid_living_area'].sum()}")
print(f"Invalid DaysOnMarket: {sold['invalid_days_on_market'].sum()}")
```

```python
sold['listing_after_close_flag'] = sold['ListingContractDate'] > sold['CloseDate']
sold['purchase_after_close_flag'] = sold['PurchaseContractDate'] > sold['CloseDate']
sold['negative_timeline_flag'] = sold['ListingContractDate'] > sold['PurchaseContractDate']

print(f"Listing after close: {sold['listing_after_close_flag'].sum()}")
print(f"Purchase after close: {sold['purchase_after_close_flag'].sum()}")
print(f"Negative timeline: {sold['negative_timeline_flag'].sum()}")
```

```python
sold['missing_coords'] = sold['Latitude'].isnull() | sold['Longitude'].isnull()
sold['zero_coords'] = (sold['Latitude'] == 0) | (sold['Longitude'] == 0)
sold['wrong_longitude'] = sold['Longitude'] > 0

print(f"Missing coordinates: {sold['missing_coords'].sum()}")
print(f"Zero coordinates: {sold['zero_coords'].sum()}")
print(f"Wrong longitude (positive): {sold['wrong_longitude'].sum()}")
```

```python
sold.to_csv('sold_cleaned.csv', index=False)
print(f"Saved sold_cleaned.csv")
print(f"Final shape: {sold.shape}")
```

```python
sold['price_ratio'] = sold['ClosePrice'] / sold['OriginalListPrice']
sold['price_per_sqft'] = sold['ClosePrice'] / sold['LivingArea']
print("Price metrics created!")
```

```python
sold['CloseDate'] = pd.to_datetime(sold['CloseDate'])
sold['PurchaseContractDate'] = pd.to_datetime(sold['PurchaseContractDate'])
sold['ListingContractDate'] = pd.to_datetime(sold['ListingContractDate'])

sold['year'] = sold['CloseDate'].dt.year
sold['month'] = sold['CloseDate'].dt.month
sold['yr_mo'] = sold['CloseDate'].dt.to_period('M')

sold['listing_to_contract_days'] = (sold['PurchaseContractDate'] - sold['ListingContractDate']).dt.days
sold['contract_to_close_days'] = (sold['CloseDate'] - sold['PurchaseContractDate']).dt.days

print("Time metrics created!")
```

```python
segment = sold.groupby('CountyOrParish').agg(
    median_close_price=('ClosePrice', 'median'),
    avg_price_per_sqft=('price_per_sqft', 'median'),
    avg_days_on_market=('DaysOnMarket', 'median'),
    avg_price_ratio=('price_ratio', 'median'),
    total_units=('ClosePrice', 'count')
).reset_index().sort_values('median_close_price', ascending=False)

print(segment.head(10))
```
